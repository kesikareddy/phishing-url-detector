```python


from urllib.parse import urlparse
import ipaddress

def detect_phishing(url):
    url = url.strip()

    if not url:
        print("Error: Please enter a URL.")
        return

    if "://" not in url:
        url = "https://" + url

    parsed = urlparse(url)
    hostname = parsed.hostname

    if parsed.scheme not in ("http", "https") or not hostname:
        print("Invalid URL. Please enter a valid HTTP/HTTPS URL.")
        return

    score = 0
    reasons = []

    if parsed.scheme != "https":
        score += 1
        reasons.append("Does not use HTTPS")

    try:
        ipaddress.ip_address(hostname)
        score += 2
        reasons.append("Uses an IP address as the host")
    except ValueError:
        pass

    if "@" in parsed.netloc:
        score += 2
        reasons.append("Contains an @ symbol in the authority")

    if len(url) > 75:
        score += 1
        reasons.append("URL is unusually long")

    keywords = ["login", "verify", "update", "password"]
    if any(word in url.lower() for word in keywords):
        score += 1
        reasons.append("Contains a sensitive keyword")

    print("\n--- URL Analysis Report ---")
    print("URL:", url)
    print("Risk score:", score)

    if score >= 3:
        print("Assessment: Potentially suspicious")
    else:
        print("Assessment: No strong warning from these checks")

    print("Reasons:")
    if reasons:
        for reason in reasons:
            print("-", reason)
    else:
        print("No configured warning patterns found.")

url = input("Enter a URL: ")
detect_phishing(url)
```
